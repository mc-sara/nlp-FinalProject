```python
import time
from publish_api import Poster, _write_ledger, get_candidates

with Poster() as poster:                        # 浏览器只开一次
    for i, c in enumerate(get_candidates(), 1):
        print(f"[{i}] 发送：{c['title']}", flush=True)
        r = poster.publish_verified(            # 必须接住返回值
            "agenttest", c["title"], c["content"],
            max_retries=3, verify_delay=3,
        )
        print(f"[{i}] {r.status}  post_id={r.post_id or '-'}  {r.elapsed}s", flush=True)
        _write_ledger(                          # 各落一个 JSON + 表格加一行
            r,
            content=c["content"],               # 不是 content
            requested_title=c["title"],         # 不是 title
            cookie=poster.cookie,
            cookie_count=poster.cookie_count,
            login_user=poster.login_user,
        )                         # 间隔自己控
```
